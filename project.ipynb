```python
import plotly.graph_objects as go
import pandas as pd

# Load dataset
data = pd.read_csv("flipkart_sample_dataset.csv")

fig = go.Figure()

# Scatter Plot (Price vs Rating)
fig.add_trace(go.Scatter(
    x=data['price'],
    y=data['rating'],
    mode='markers',
    name='Scatter',
    visible=True
))

# Bar Chart (Products per Category)
category_count = data['category'].value_counts()
fig.add_trace(go.Bar(
    x=category_count.index,
    y=category_count.values,
    name='Bar',
    visible=False
))

# Histogram (Price Distribution)
fig.add_trace(go.Histogram(
    x=data['price'],
    name='Histogram',
    visible=False
))

# Box Plot (Rating Distribution by Brand)
fig.add_trace(go.Box(
    x=data['brand'],
    y=data['rating'],
    name='Box Plot',
    visible=False
))

# Line Chart (Average Price by Brand)
avg_price = data.groupby("brand")["price"].mean()

fig.add_trace(go.Scatter(
    x=avg_price.index,
    y=avg_price.values,
    mode='lines+markers',
    name='Line Chart',
    visible=False
))

# Dropdown Menu
fig.update_layout(
    title="Flipkart Product Data Analysis",
    updatemenus=[
        dict(
            buttons=list([
                dict(label="Scatter Plot",
                     method="update",
                     args=[{"visible":[True, False, False, False, False]},
                           {"title":"Price vs Rating"}]),

                dict(label="Bar Chart",
                     method="update",
                     args=[{"visible":[False, True, False, False, False]},
                           {"title":"Products per Category"}]),

                dict(label="Histogram",
                     method="update",
                     args=[{"visible":[False, False, True, False, False]},
                           {"title":"Price Distribution"}]),

                dict(label="Box Plot",
                     method="update",
                     args=[{"visible":[False, False, False, True, False]},
                           {"title":"Rating Distribution by Brand"}]),

                dict(label="Line Chart",
                     method="update",
                     args=[{"visible":[False, False, False, False, True]},
                           {"title":"Average Price by Brand"}]),
            ]),
            direction="down"
        )
    ]
)

fig.show()
```
