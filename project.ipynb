```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv("products.csv")
df.head()
```

```python
print(df.shape)
print(df.columns)
print(df.info())
```

```python
brand_price = df.groupby("brand")["price"].mean().reset_index()

fig = px.bar(
    brand_price,
    x="brand",
    y="price",
    color="brand",
    title="Average Price by Brand"
)
fig.update_layout( plot_bgcolor="black",paper_bgcolor="black",font=dict(color="white")   )
fig.show()
```

```python
fig = px.scatter(
    df,
    x="price",
    y="rating",
    color="brand",
    size="reviews",
    hover_data=["product_name"],
    title="Price vs Rating Analysis"
)
fig.update_layout( plot_bgcolor="black",paper_bgcolor="black",font=dict(color="white")   )
fig.show()
```

```python
category_count = df["category"].value_counts().reset_index()
category_count.columns = ["category", "count"]

fig = px.pie(
    category_count,
    values="count",
    names="category",
    title="Product Category Distribution"
)
fig.update_layout( plot_bgcolor="black",paper_bgcolor="black",font=dict(color="white")   )
fig.show()
```

```python
fig = px.histogram(df,x="rating",nbins=20,color_discrete_sequence=["cyan"],title="Ratings Distribution")
fig.update_traces(marker_line_color="blue",marker_line_width=2)
fig.update_layout( plot_bgcolor="black",paper_bgcolor="black",font=dict(color="white")   )
fig.show()
```

```python
fig = px.scatter(df,x="discount_percent",y="price",size="reviews",
                 color="brand",hover_data=["product_name"],title="Discount vs Price Analysis")
fig.update_layout( plot_bgcolor="black",paper_bgcolor="black",font=dict(color="white")   )

fig.show()
```

```python
fig = go.Figure()

brands = df["brand"].unique()

for brand in brands:
    brand_df = df[df["brand"] == brand]

    colors = px.colors.qualitative.Plotly

    fig.add_trace(
        go.Bar(
            x=brand_df["product_name"],
            y=brand_df["price"],
            name=brand,
            marker=dict(
                color=colors * 20,
                line=dict(color="blue", width=2)
            ),
            hoverlabel=dict(
                bgcolor="maroon",
                font_color="white"
            ),
            visible=(brand == brands[0])
        )
    )

buttons = []

for i, brand in enumerate(brands):
    visible = [False] * len(brands)
    visible[i] = True

    buttons.append(
        dict(
            label=brand,
            method="update",
            args=[
                {"visible": visible},
                {"title": f"Product Prices for {brand}"}
            ]
        )
    )

fig.update_layout(

    title="Product Price Analysis by Brand",

    xaxis_title="Product Name",
    yaxis_title="Price",

    plot_bgcolor="black",
    paper_bgcolor="black",

    font=dict(color="white"),

    xaxis=dict(tickangle=-45),

    updatemenus=[
        dict(
            active=0,
            buttons=buttons,
            bgcolor="black",
            bordercolor="white",
            font=dict(color="orange")
        )
    ]
)

fig.show()
```

```python
brand_sales = df.groupby("brand")["reviews"].sum().reset_index()

base_colors = px.colors.qualitative.Set2
colors = (base_colors * 10)[:len(brand_sales)]

fig = go.Figure()

fig.add_trace(go.Bar(
    x=brand_sales["brand"],
    y=brand_sales["reviews"],
    marker=dict(color=colors),
    visible=True
))

fig.add_trace(go.Scatter(
    x=brand_sales["brand"],
    y=brand_sales["reviews"],
    mode="lines+markers",
    marker=dict(color=colors),
    visible=False
))

fig.add_trace(go.Pie(
    labels=brand_sales["brand"],
    values=brand_sales["reviews"],
    marker=dict(colors=colors),
    visible=False
))

fig.add_trace(go.Scatter(
    x=brand_sales["brand"],
    y=brand_sales["reviews"],
    mode="markers",
    marker=dict(size=15,color=colors),
    visible=False
))

fig.add_trace(go.Pie(
    labels=brand_sales["brand"],
    values=brand_sales["reviews"],
    hole=0.5,
    marker=dict(colors=colors),
    visible=False
))

fig.add_trace(go.Violin(
    x=df["brand"],
    y=df["reviews"],
    line_color="lightblue",
    fillcolor="rgba(0,176,246,0.4)",
    visible=False
))

fig.update_layout(
    updatemenus=[
        dict(
            buttons=[
                dict(label="Bar", args=[{"visible":[True,False,False,False,False,False]}], method="update"),
                dict(label="Line", args=[{"visible":[False,True,False,False,False,False]}], method="update"),
                dict(label="Pie", args=[{"visible":[False,False,True,False,False,False]}], method="update"),
                dict(label="Scatter", args=[{"visible":[False,False,False,True,False,False]}], method="update"),
                dict(label="Donut", args=[{"visible":[False,False,False,False,True,False]}], method="update"),
                dict(label="Violin", args=[{"visible":[False,False,False,False,False,True]}], method="update")
            ],
            direction="down",
            bgcolor="black",
            bordercolor="white",
            font=dict(color="orange")
        )
    ],
    title="Sales Analysis by Brand",
    plot_bgcolor="black",
    paper_bgcolor="black",
    font=dict(color="white")
)

fig.show()
```
